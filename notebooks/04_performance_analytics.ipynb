# Performance Analytics

This notebook evaluates mutual fund performance using CAGR,
Volatility, Sharpe Ratio, Sortino Ratio, Alpha, Beta,
Maximum Drawdown and Benchmark Comparison.

Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_style("whitegrid")
```

Load Outputs

```python
processed_dir = Path("../data/processed")

scorecard = pd.read_csv(
    processed_dir /
    "fund_scorecard.csv"
)

alpha_beta = pd.read_csv(
    processed_dir /
    "alpha_beta.csv"
)

fund_metrics = pd.read_csv(
    processed_dir /
    "fund_metrics.csv"
)
```

CAGR Distribution

```python
plt.figure(figsize=(10,5))

sns.histplot(
    scorecard["cagr_3yr_pct"],
    kde=True
)

plt.title(
    "3-Year CAGR Distribution"
)

plt.show()
```

Sharpe Ratio Ranking

```python
top10 = (
    scorecard
    .sort_values(
        "sharpe_ratio_y",
        ascending=False
    )
    .head(10)
)

plt.figure(figsize=(10,5))

sns.barplot(
    data=top10,
    x="sharpe_ratio_y",
    y="scheme_name"
)

plt.title(
    "Top 10 Funds by Sharpe Ratio"
)

plt.show()
```

Sortino Ratio Ranking

```python
top10 = (
    scorecard
    .sort_values(
        "sortino_ratio_y",
        ascending=False
    )
    .head(10)
)

plt.figure(figsize=(10,5))

sns.barplot(
    data=top10,
    x="sortino_ratio_y",
    y="scheme_name"
)

plt.show()
```

Alpha Distribution

```python
sns.histplot(
    scorecard["alpha"],
    kde=True
)

plt.title(
    "Alpha Distribution"
)

plt.show()
```

Beta Distribution

```python
sns.histplot(
    scorecard["beta"],
    kde=True
)

plt.title(
    "Beta Distribution"
)

plt.show()
```

Volatility vs Return

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=scorecard,
    x="volatility_pct",
    y="cagr_3yr_pct",
    hue="category"
)

plt.show()
```

Max Drawdown Analysis

```python
top10 = (
    scorecard
    .sort_values(
        "max_drawdown_pct_y"
    )
    .head(10)
)

sns.barplot(
    data=top10,
    x="max_drawdown_pct_y",
    y="scheme_name"
)

plt.show()
```

Fund Score Distribution

```python
sns.histplot(
    scorecard["fund_score"],
    kde=True
)

plt.show()
```

Top 10 Funds

```python
top10 = (
    scorecard
    .sort_values(
        "fund_score",
        ascending=False
    )
    .head(10)
)

sns.barplot(
    data=top10,
    x="fund_score",
    y="scheme_name"
)

plt.show()
```

Expense Ratio vs Return

```python
sns.scatterplot(
    data=scorecard,
    x="expense_ratio_pct",
    y="cagr_3yr_pct"
)

plt.show()
```
